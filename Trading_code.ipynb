```python
import pandas as pd

# Reemplaza con tu enlace raw
url = "https://github.com/asantibane-star/datascience.trading/raw/refs/heads/main/base_usdclp_limpia_v2.xlsx"

# Si el archivo tiene varias hojas, especifica sheet_name
datos = pd.read_excel(url, sheet_name = 'datos', engine="openpyxl")
datos['fecha'] = pd.to_datetime(datos['fecha'])
datos = datos.set_index('fecha').sort_index()

datos.describe()
```

```python
"""Construye la base X (variables) e y (objetivo) a partir de base_usdclp_limpia_v2.xlsx.

Regla general: el modelo no recibe niveles de precios (el dólar en 2026 está muy por sobre lo visto
en el entrenamiento), sino retornos, diferencias, distancias a promedios e indicadores acotados.
Todo usa solo información disponible al cierre del día t.
"""
import numpy as np

# ---------------------------------------------------------------------------
# 1. Leer la base limpia
# ---------------------------------------------------------------------------

# Filas cuyo día hábil anterior fue eliminado: ahí una variación diaria cruzaría varios días
salto = datos['fila_contigua'] == 0

# Precios del dólar
c = datos['USDCLP Curncy|PX_LAST']   # cierre
o = datos['USDCLP Curncy|PX_OPEN']   # apertura
h = datos['USDCLP Curncy|PX_HIGH']   # máximo
l = datos['USDCLP Curncy|PX_LOW']    # mínimo

X = pd.DataFrame(index=datos.index)

# ---------------------------------------------------------------------------
# 2. Precios e índices -> retorno logarítmico diario
# ---------------------------------------------------------------------------
precios = [
    'HG1 Comdty|PX_LAST', 'LMCADS03 Comdty|PX_LAST', 'LMAHDS03 Comdty|PX_LAST',   # cobre y aluminio
    'CO1 Comdty|PX_LAST', 'XAU Curncy|PX_LAST', 'BCOM Index|PX_LAST',             # Brent, oro, commodities
    'DXY Curncy|PX_LAST', 'BBDXY Index|PX_LAST', 'EURUSD Curncy|PX_LAST',          # dólar global
    'USDJPY Curncy|PX_LAST', 'USDCNH',
    'USDBRL Curncy|PX_LAST', 'USDMXN Curncy|PX_LAST', 'USDCOP Curncy|PX_LAST',     # monedas emergentes
    'USDPEN Curncy|PX_LAST', 'USDZAR Curncy|PX_LAST',
    'SPX Index|PX_LAST', 'MXEF Index|PX_LAST', 'SHCOMP Index|PX_LAST',             # bolsas
    'CHUF Index|PX_LAST',                                                          # UF
]
for col in precios:
    X['r_' + col.split()[0]] = np.log(datos[col]).diff()

# ---------------------------------------------------------------------------
# 3. Tasas, puntos forward, volatilidades y posiciones -> diferencia diaria
# ---------------------------------------------------------------------------
tasas = [
    'CHN1W Curncy|PX_LAST', 'CHN1M Curncy|PX_LAST', 'CHN3M Curncy|PX_LAST',        # puntos forward
    'CHN6M Curncy|PX_LAST', 'CHN12M Curncy|PX_LAST',
    'CHNI1M Curncy|PX_LAST', 'CHNI3M Curncy|PX_LAST', 'CHNI6M Curncy|PX_LAST',     # tasas implícitas NDF
    'CHSWP1 Curncy|PX_LAST', 'CHSWP2 Curncy|PX_LAST', 'CHSWP5 Curncy|PX_LAST',     # swaps en pesos
    'CHSWP10 Curncy|PX_LAST', 'CHSWC2 Curncy|PX_LAST',
    'TPM', 'TASA_CAMARA_NOM', 'TASA_USD_ON',                                       # tasas cortas
    'USGG3M Index|PX_LAST', 'USGG2YR Index|PX_LAST', 'USGG10YR Index|PX_LAST',     # Tesoro EE.UU.
    'VIX Index|PX_LAST', 'MOVE Index|PX_LAST', 'JPMVXYEM Index|PX_LAST',           # volatilidades
    'USDCLPV1M Curncy|PX_LAST', 'USDCLPV3M Curncy|PX_LAST',
    'POS_DERIV_OFFSHORE', 'POS_DERIV_LOCAL',                                       # posiciones en derivados
]
for col in tasas:
    X['d_' + col.split()[0]] = datos[col].diff()

X.loc[salto] = np.nan   # no calcular variaciones a través de días eliminados

# ---------------------------------------------------------------------------
# 4. Nivel del dólar, corregido para que no dependa del precio
# ---------------------------------------------------------------------------
log_c = np.log(c)
X['usd_vs_media50'] = log_c - log_c.rolling(50).mean()       # % sobre/bajo su promedio de 50 días
X['usd_vs_media200'] = log_c - log_c.rolling(200).mean()     # % sobre/bajo su promedio de 200 días
X['usd_z252'] = (log_c - log_c.rolling(252).mean()) / log_c.rolling(252).std()   # desvíos vs el último año

# ---------------------------------------------------------------------------
# 5. Retornos y volatilidad del dólar
# ---------------------------------------------------------------------------
ret = log_c.diff()
ret[salto] = np.nan
X['usd_ret1'] = ret                                   # retorno de hoy
X['usd_ret1_ayer'] = ret.shift(1)                     # retorno de ayer
X['usd_ret5'] = log_c.diff(5)                         # retorno de la semana
X['usd_ret20'] = log_c.diff(20)                       # retorno del mes
X['usd_vol20'] = ret.rolling(20, min_periods=15).std()   # volatilidad del último mes
X['usd_rango'] = (h - l) / c                          # amplitud del día
X['usd_cuerpo'] = (c - o) / c                         # cierre contra apertura
X['usd_vs_observado'] = np.log(c / datos['CLFXDOOB Index|PX_LAST'])   # el observado de t es el spot de t-1

# ---------------------------------------------------------------------------
# 6. Indicadores técnicos del dólar
# ---------------------------------------------------------------------------
# Williams %R (14 días): dónde cerró dentro del rango de 14 días. 0 = máximo, -100 = mínimo
maximo14 = h.rolling(14).max()
minimo14 = l.rolling(14).min()
X['williams_r'] = -100 * (maximo14 - c) / (maximo14 - minimo14)
X['williams_sobrecompra'] = (X['williams_r'] > -20).astype(int)
X['williams_sobreventa'] = (X['williams_r'] < -80).astype(int)

# RSI (14 días, suavizado de Wilder): fuerza de las alzas contra las bajas. Va de 0 a 100
cambio = c.diff()
alzas = cambio.clip(lower=0).ewm(alpha=1 / 14, adjust=False).mean()
bajas = (-cambio.clip(upper=0)).ewm(alpha=1 / 14, adjust=False).mean()
X['rsi'] = 100 - 100 / (1 + alzas / bajas)
X['rsi_sobrecompra'] = (X['rsi'] > 70).astype(int)
X['rsi_sobreventa'] = (X['rsi'] < 30).astype(int)

# Bandas de Bollinger (20 días, 2 desviaciones)
media20 = c.rolling(20).mean()
desv20 = c.rolling(20).std()
boll_sup = media20 + 2 * desv20
boll_inf = media20 - 2 * desv20
X['bollinger_pct_b'] = (c - boll_inf) / (boll_sup - boll_inf)   # 0 = banda inferior, 1 = banda superior
X['bollinger_ancho'] = (boll_sup - boll_inf) / media20           # ancho de la banda en % del precio
X['bollinger_sobre'] = (c > boll_sup).astype(int)
X['bollinger_bajo'] = (c < boll_inf).astype(int)

# Canal de Keltner (media exponencial de 20 días, 2 veces el ATR de 20 días)
rango_verdadero = pd.concat([h - l, (h - c.shift(1)).abs(), (l - c.shift(1)).abs()], axis=1).max(axis=1)
atr20 = rango_verdadero.rolling(20).mean()
centro = c.ewm(span=20, adjust=False).mean()
X['keltner_pos'] = (c - centro) / (2 * atr20)   # -1 = banda inferior, +1 = banda superior
X['keltner_ancho'] = 4 * atr20 / centro         # ancho del canal en % del precio
X['keltner_sobre'] = (X['keltner_pos'] > 1).astype(int)
X['keltner_bajo'] = (X['keltner_pos'] < -1).astype(int)

# ---------------------------------------------------------------------------
# 7. Diferenciales (en nivel: oscilan en un rango estable)
# ---------------------------------------------------------------------------
X['carry'] = datos['TPM'] - datos['TASA_USD_ON']                                        # tasa Chile - tasa EE.UU.
X['pendiente_cl'] = datos['CHSWP10 Curncy|PX_LAST'] - datos['CHSWP1 Curncy|PX_LAST']    # curva en pesos
X['infl_implicita_2y'] = datos['CHSWP2 Curncy|PX_LAST'] - datos['CHSWC2 Curncy|PX_LAST']
X['pendiente_us'] = datos['USGG10YR Index|PX_LAST'] - datos['USGG2YR Index|PX_LAST']   # curva EE.UU.
X['prima_vol_clp'] = datos['USDCLPV1M Curncy|PX_LAST'] - X['usd_vol20'] * np.sqrt(252) * 100   # implícita - realizada
X['vix'] = datos['VIX Index|PX_LAST']

# ---------------------------------------------------------------------------
# 8. Macro (ya son tasas de variación) y calendario
# ---------------------------------------------------------------------------
for col in ['CLIMMOMS Index', 'CLIMYOYN Index', 'CLINNSMO Index', 'CLINNSYO Index', 'CHUETOTL Index']:
    X[col.split()[0]] = datos[col]
X['dia_semana'] = datos.index.dayofweek

# ---------------------------------------------------------------------------
# 9. Objetivo y filas utilizables
# ---------------------------------------------------------------------------
completas = X.notna().all(axis=1)           # descarta el primer año (ventana de 252 días) y los saltos
x_hoy = X[completas].iloc[[-1]]             # último cierre: con esta fila se calcula la señal de mañana

utiles = completas & datos['Y'].notna()
X = X[utiles]
y = datos.loc[utiles, 'Y'].astype(int)
ret_siguiente = datos.loc[utiles, 'ret_siguiente']    # para calcular el resultado de la estrategia

if __name__ == '__main__':
    print('X:', X.shape, '| periodo:', X.index.min().date(), 'a', X.index.max().date())
    print('Proporción de alzas (y = 1):', round(y.mean(), 3))
    print('Señal de hoy con datos del', x_hoy.index[0].date())
```

```python
# =============================================================================
# 2.1 Parámetros, carry y resultado de una estrategia
# =============================================================================
import lightgbm as lgb
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score

COSTO = 0              # costo por unidad de nocional transada (3 pb). Reemplazar por el del código de evaluación
INICIO_TEST = '2023-01-01'  # tramo fuera de muestra: nada desde aquí se usa para elegir parámetros
REENTRENO = 21              # cada ~1 mes se reentrena con todo lo anterior (walk-forward)
MINIMO = 756                # 3 años de historia antes de la primera predicción

# Un largo en dólar paga el carry (tasa en pesos - tasa en dólares) por los días corridos hasta el siguiente día hábil
carry = ((datos['TASA_CAMARA_NOM'] - datos['TASA_USD_ON']) / 100 * datos['dias_corridos_siguiente'] / 360).loc[X.index]
ret_neto = ret_siguiente - carry        # lo que gana una exposición +1 antes de costos
y_neto = (ret_neto > 0).astype(int)     # objetivo: ¿le conviene estar comprado en dólar mañana?
test = X.index >= INICIO_TEST

def pnl(pos, costo=COSTO):
    """Resultado diario de una serie de exposiciones (-3 a 3), neto de carry y de costos."""
    pos = pd.Series(pos, index=X.index).fillna(0)
    return pos * ret_neto - costo * pos.diff().abs().fillna(pos.abs())

def sharpe(r):
    return r.mean() / r.std() * np.sqrt(252)

print(f'Desarrollo: {X.index[MINIMO].date()} a {INICIO_TEST} | Fuera de muestra: {INICIO_TEST} a {X.index[-1].date()} ({test.sum()} días)')
print(f'Carry medio anual (pesos - dólar): {(carry / datos["dias_corridos_siguiente"].loc[X.index]).mean() * 360:.2%}')
```

```python
# =============================================================================
# 2.2 Ensamble primario: LightGBM + logística, reentrenado cada mes
# =============================================================================
# Asegurar que las variables interactivas existan en el DataFrame X
if 'd_CHN1W pendiente_us' not in X.columns:
    X['d_CHN1W pendiente_us'] = X['d_CHN1W'] * X['pendiente_us']
    X['d_CHNI3M CLIMYOYN'] = X['d_CHNI3M'] * X['CLIMYOYN']
    X['usd_rango dia_semana'] = X['usd_rango'] * X['dia_semana']
    X['r_USDMXN d_USGG3M'] = X['r_USDMXN'] * X['d_USGG3M']
    X['dia_semana^2'] = X['dia_semana'] ** 2
    X['d_CHN6M CLIMMOMS'] = X['d_CHN6M'] * X['CLIMMOMS']
    X['d_CHNI3M bollinger_pct_b'] = X['d_CHNI3M'] * X['bollinger_pct_b']
    X['d_USGG10YR CLIMMOMS'] = X['d_USGG10YR'] * X['CLIMMOMS']
    X['d_JPMVXYEM d_POS_DERIV_LOCAL'] = X['d_JPMVXYEM'] * X['d_POS_DERIV_LOCAL']
    X['d_USGG10YR d_POS_DERIV_LOCAL'] = X['d_USGG10YR'] * X['d_POS_DERIV_LOCAL']
    X['bollinger_pct_b dia_semana'] = X['bollinger_pct_b'] * X['dia_semana']
    X['r_USDMXN bollinger_ancho'] = X['r_USDMXN'] * X['bollinger_ancho']
    X['r_USDMXN usd_vol20'] = X['r_USDMXN'] * X['usd_vol20']
    X['r_XAU pendiente_us'] = X['r_XAU'] * X['pendiente_us']

variables = [
    'usd_ret1', 'usd_ret5', 'usd_ret20',                     # momentum del dólar
    'usd_vs_media50', 'usd_vs_media200', 'usd_z252',         # memoria de largo plazo, sin usar el nivel
    'usd_vol20', 'prima_vol_clp', 'vix',                     # régimen de volatilidad
    'usd_cuerpo', 'usd_vs_observado',                        # cómo cerró el día
    'r_HG1', 'r_DXY', 'r_USDBRL', 'r_USDMXN', 'r_USDCNH',    # cobre, dólar global y monedas de la región
    'r_SPX', 'r_MXEF',                                       # apetito por riesgo
    'carry', 'd_CHN1M', 'd_TPM',                             # diferencial de tasas y puntos forward
    'd_CHN1W pendiente_us',
    'd_CHNI3M CLIMYOYN',
    'usd_rango dia_semana',
    'r_USDMXN d_USGG3M',
    'dia_semana^2',
    'd_CHN6M CLIMMOMS',
    'd_CHNI3M bollinger_pct_b',
    'd_USGG10YR CLIMMOMS',
    'd_JPMVXYEM d_POS_DERIV_LOCAL',
    'd_USGG10YR d_POS_DERIV_LOCAL',
    'bollinger_pct_b dia_semana',
    'r_USDMXN bollinger_ancho',
    'r_USDMXN usd_vol20',
    'r_XAU pendiente_us'
]

def modelos():
    return {
        'arbol': lgb.LGBMClassifier(n_estimators=200, learning_rate=0.01, num_leaves=5, min_child_samples=100,
                                    subsample=0.7, subsample_freq=1, colsample_bytree=0.5, reg_lambda=10,
                                    random_state=0, verbose=-1, n_jobs=1),
        'logit': make_pipeline(StandardScaler(), LogisticRegression(C=0.01, max_iter=3000)),
    }

def ajustar(modelo, x, y_, peso):
    """Entrena ponderando cada día por |retorno|: importan los días que mueven el resultado."""
    peso = (peso / peso.mean()).values
    if hasattr(modelo, 'steps'):
        return modelo.fit(x, y_, logisticregression__sample_weight=peso)
    return modelo.fit(x, y_, sample_weight=peso)

def walk_forward(x, y_, peso):
    """Probabilidades fuera de muestra: para predecir el día i solo se usan los días anteriores a i."""
    P = pd.DataFrame(index=x.index, columns=['arbol', 'logit'], dtype=float)
    for i in range(MINIMO, len(x), REENTRENO):
        for nombre, modelo in modelos().items():
            ajustar(modelo, x.iloc[:i], y_.iloc[:i], peso.iloc[:i])
            P.iloc[i:i + REENTRENO, P.columns.get_loc(nombre)] = modelo.predict_proba(x.iloc[i:i + REENTRENO])[:, 1]
    return P

P = walk_forward(X[variables], y_neto, ret_neto.abs())
P['p'] = P[['arbol', 'logit']].mean(axis=1)    # probabilidad de que un largo en dólar gane neto de carry

dev = P['p'].notna() & ~test
for nombre in ['arbol', 'logit', 'p']:
    print(f'{nombre:6s} AUC desarrollo {roc_auc_score(y_neto[dev], P.loc[dev, nombre]):.3f} | '
          f'AUC fuera de muestra {roc_auc_score(y_neto[test], P.loc[test, nombre]):.3f}')
```

```python
# =============================================================================
# 2.3 Decisión: dirección, convicción y exposición (-3, -1, 0, 1, 3)
# =============================================================================
direccion = np.sign(P['p'] - 0.5)
acuerdo = np.sign(P['arbol'] - 0.5) == np.sign(P['logit'] - 0.5)
conviccion = (P['p'] - 0.5).abs()

def posicion(u1, u3):
    """0 si los modelos discrepan o la convicción < u1; 1 si está entre u1 y u3; 3 si supera u3."""
    nivel = np.where(conviccion >= u3, 3, np.where(conviccion >= u1, 1, 0))
    return (direccion * nivel * acuerdo).fillna(0) + 0.0   # + 0.0 evita el -0.0

# Umbrales elegidos SOLO con el período de desarrollo (percentiles de la convicción, incluye "nunca 3")
q = conviccion[dev].quantile
grilla = [(q(a), q(b) if b < 1 else np.inf) for a in [0, 0.2, 0.4, 0.6] for b in [0.8, 0.9, 0.95, 1]]
u1, u3 = max(grilla, key=lambda u: sharpe(pnl(posicion(*u))[dev]))
pos = posicion(u1, u3)

print(f'Umbral para operar: convicción >= {u1:.4f} | umbral para exposición 3: {u3:.4f}')
print(f'Sharpe en desarrollo: {sharpe(pnl(pos)[dev]):.2f}')
print('Exposiciones fuera de muestra:', pos[test].value_counts().sort_index().to_dict())
```

```python
# =============================================================================
# 2.4 Meta-labeling del informe (se deja como comparación)
# =============================================================================
# Un segundo modelo estima si seguir al primario habrá ganado neto de costos y decide si operar
acierto = (direccion * ret_neto > COSTO).astype(int)
M = pd.DataFrame({'conviccion': conviccion, 'desacuerdo': (P['arbol'] - P['logit']).abs(),
                  'vol': X['usd_vol20'], 'vix': X['vix'], 'prima_vol': X['prima_vol_clp'],
                  'momentum': X['usd_ret20'] * direccion, 'carry': X['carry'] * direccion})

p_meta = pd.Series(np.nan, index=X.index)
inicio = np.flatnonzero(P['p'].notna())[0]
for i in range(inicio + 504, len(X), REENTRENO):     # 2 años de predicciones del primario para aprender
    meta = make_pipeline(StandardScaler(), LogisticRegression(C=0.1, max_iter=2000))
    meta.fit(M.iloc[inicio:i], acierto.iloc[inicio:i])
    p_meta.iloc[i:i + REENTRENO] = meta.predict_proba(M.iloc[i:i + REENTRENO])[:, 1]

pos_meta = (direccion * (p_meta >= p_meta[dev].median())).fillna(0)
dev_meta = p_meta.notna() & ~test
print(f'AUC meta-modelo desarrollo {roc_auc_score(acierto[dev_meta], p_meta[dev_meta]):.3f} | '
      f'fuera de muestra {roc_auc_score(acierto[test], p_meta[test]):.3f}  (0.5 = azar)')
```

# 3. Robustez exigida por las bases

Todo se mide en el tramo fuera de muestra (desde `INICIO_TEST`), que no se usó para elegir variables, hiperparámetros ni umbrales. Se compara contra las tres referencias de las bases: **comprar siempre**, **no operar nunca** y **operar siempre con exposición uno** (seguir la dirección del modelo con ±1 todos los días).

```python
# =============================================================================
# 3.1 Rentabilidad fuera de muestra, antes y después de costos
# =============================================================================
estrategias = {
    'Modelo': pos,
    'Comprar siempre': pd.Series(1.0, index=X.index),
    'No operar nunca': pd.Series(0.0, index=X.index),
    'Exposición uno (±1)': direccion.fillna(0),
    'Meta-labeling (informe)': pos_meta,
}

def resumen(p):
    neto, bruto = pnl(p)[test], pnl(p, costo=0)[test]
    curva = (1 + neto).cumprod()
    operados = p[test] != 0
    return {
        'Ret. sin costos': (1 + bruto).prod() - 1,
        'Ret. neto': curva.iloc[-1] - 1,
        'Ret. neto anual': curva.iloc[-1] ** (252 / len(neto)) - 1,
        'Sharpe neto': sharpe(neto) if neto.std() > 0 else 0,
        'Máx. caída': (curva / curva.cummax() - 1).min(),
        '% días operando': operados.mean(),
        'Acierto días operados': (np.sign(p[test]) == np.sign(ret_neto[test]))[operados].mean(),
    }

tabla = pd.DataFrame({k: resumen(v) for k, v in estrategias.items()}).T
tabla.style.format('{:.2%}').format('{:.2f}', subset=['Sharpe neto'])
```

```python
# =============================================================================
# 3.2 Test de Diebold-Mariano
# =============================================================================
from scipy import stats

def diebold_mariano(perdida_modelo, perdida_ref, rezagos=5):
    """DM con varianza Newey-West y corrección de Harvey-Leybourne-Newbold.
    H0: igual pérdida esperada. H1: el modelo tiene menor pérdida (p-valor de una cola)."""
    d = (perdida_modelo - perdida_ref).values
    n, dc = len(d), d - d.mean()
    var = dc @ dc / n + 2 * sum((1 - k / (rezagos + 1)) * (dc[k:] @ dc[:-k]) / n for k in range(1, rezagos + 1))
    dm = d.mean() / np.sqrt(var / n) * np.sqrt((n - 1) / n)
    return dm, stats.t.cdf(dm, n - 1)

# a) Calidad de la predicción: Brier score de la probabilidad contra pronósticos ingenuos
brier = lambda prob: (prob - y_neto[test]) ** 2
ref_prob = {'Moneda (p = 0.5)': pd.Series(0.5, index=X.index)[test],
            'Frecuencia histórica de alzas': y_neto.expanding().mean().shift(1)[test]}
filas = {f'Brier vs {k}': diebold_mariano(brier(P.loc[test, 'p']), brier(v)) for k, v in ref_prob.items()}

# b) Valor económico: pérdida = -resultado diario neto, contra las tres referencias
for k in ['Comprar siempre', 'No operar nunca', 'Exposición uno (±1)']:
    filas[f'Resultado vs {k}'] = diebold_mariano(-pnl(pos)[test], -pnl(estrategias[k])[test])

pd.DataFrame(filas, index=['DM', 'p-valor']).T.round(3)
```

```python
# =============================================================================
# 3.3 Bootstrap no paramétrico por bloques sobre el resultado de la estrategia
# =============================================================================
def bootstrap(diferencia, n=5000, bloque=20, semilla=0):
    """Remuestrea bloques de 20 días (conserva la autocorrelación).
    Devuelve el p-valor de H0: media <= 0 y el intervalo al 95% del exceso anual."""
    d = diferencia.values
    T = len(d)
    rng = np.random.default_rng(semilla)
    inicios = rng.integers(0, T, size=(n, T // bloque + 1))
    idx = ((inicios[:, :, None] + np.arange(bloque)).reshape(n, -1)[:, :T]) % T
    medias = d[idx].mean(axis=1)
    p_valor = np.mean(medias - d.mean() >= d.mean())       # distribución centrada en H0
    return p_valor, *(np.percentile(medias, [2.5, 97.5]) * 252)

filas = {}
for k in ['Comprar siempre', 'No operar nunca', 'Exposición uno (±1)']:
    filas[f'Modelo - {k}'] = bootstrap(pnl(pos)[test] - pnl(estrategias[k])[test])
pd.DataFrame(filas, index=['p-valor', 'IC 2.5% anual', 'IC 97.5% anual']).T.round(4)
```

```python
# =============================================================================
# 3.4 Controles adicionales: permutación, costos, años y calibración
# =============================================================================
# a) Permutación de etiquetas (control 1 del informe): si se entrena con etiquetas desordenadas,
#    el AUC fuera de muestra debe caer a ~0.5. Si no cae, hay fuga en las variables.
N_PERM = 10
rng = np.random.default_rng(1)
auc_real = roc_auc_score(y_neto[test], P.loc[test, 'p'])
auc_perm = []
for _ in range(N_PERM):
    orden = rng.permutation(len(X))
    Pp = walk_forward(X[variables], y_neto.iloc[orden].set_axis(X.index), ret_neto.abs().iloc[orden].set_axis(X.index))
    auc_perm.append(roc_auc_score(y_neto[test], Pp.loc[test].mean(axis=1)))
print(f'AUC real {auc_real:.3f} | AUC con etiquetas desordenadas {np.mean(auc_perm):.3f} ± {np.std(auc_perm):.3f} '
      f'| p-valor {np.mean(np.array(auc_perm) >= auc_real):.2f}')

# b) Sensibilidad a costos
print('\nRetorno neto fuera de muestra según el costo por unidad transada:')
for c in [0, COSTO, 2 * COSTO, 4 * COSTO]:
    print(f'  {c * 1e4:4.0f} pb: {(1 + pnl(pos, c)[test]).prod() - 1:7.2%}')

# c) Resultado por año (desarrollo y fuera de muestra): ¿depende de un solo año?
r = pnl(pos)[P['p'].notna()]
print('\nResultado por año:', r.groupby(r.index.year).sum().map('{:.2%}'.format).to_dict())

# d) Calibración: probabilidad predicha vs frecuencia observada de alzas netas, por quintil
cal = pd.DataFrame({'p predicha': P.loc[test, 'p'], 'frecuencia real': y_neto[test]})
cal.groupby(pd.qcut(cal['p predicha'], 5)).mean().round(3)
```

```python
# =============================================================================
# 4.1 Señal del día: se entrena con toda la historia y se predice con el último cierre
# =============================================================================
finales = {k: ajustar(m, X[variables], y_neto, ret_neto.abs()) for k, m in modelos().items()}
p_arbol, p_logit = (finales[k].predict_proba(x_hoy[variables])[0, 1] for k in ['arbol', 'logit'])
p_hoy = (p_arbol + p_logit) / 2
conv_hoy = abs(p_hoy - 0.5)
nivel_hoy = 3 if conv_hoy >= u3 else 1 if conv_hoy >= u1 else 0
pos_hoy = int(np.sign(p_hoy - 0.5) * nivel_hoy * (np.sign(p_arbol - 0.5) == np.sign(p_logit - 0.5)))

print(f'Datos al cierre del {x_hoy.index[0].date()}')
print(f'Dirección esperada: {"SUBE" if p_hoy > 0.5 else "BAJA"}')
print(f'Probabilidad de alza neta de carry: {p_hoy:.1%} (árbol {p_arbol:.1%}, logística {p_logit:.1%})')
print(f'Posición sugerida: {pos_hoy:+d}')
```

```python
# =============================================================================
# 4.3 Lectura de la convicción: jornadas de máxima exposición y de abstención
# =============================================================================
t = pd.DataFrame({'pos': pos, 'p': P['p'], 'acuerdo': acuerdo, 'ret_neto': ret_neto, 'resultado': pnl(pos)})[test]
t['nivel'] = t['pos'].abs()
t['acierto'] = np.sign(t['pos']) == np.sign(t['ret_neto'])
t['motivo'] = np.where(t['nivel'] > 0, 'opera', np.where(~t['acuerdo'], 'modelos discrepan', 'convicción baja'))

lectura = t.groupby('nivel').agg(dias=('pos', 'size'),
                                 acierto=('acierto', 'mean'),
                                 resultado_medio_pb=('resultado', lambda r: r.mean() * 1e4),
                                 movimiento_medio_pb=('ret_neto', lambda r: r.abs().mean() * 1e4))
print('Por nivel de exposición (acierto no aplica al nivel 0):')
display(lectura.round(2))

print('Motivo de abstención:', t.loc[t['nivel'] == 0, 'motivo'].value_counts().to_dict())
print(f'Lo que se dejó de ganar o perder al abstenerse con exposición uno: '
      f'{(direccion[test][t["nivel"] == 0] * ret_neto[test][t["nivel"] == 0]).sum():.2%}')

nivel_max = t['nivel'].max()
print(f'\nÚltimas jornadas con exposición máxima ({nivel_max:.0f}):')
display(t[t['nivel'] == nivel_max].tail(10)[['pos', 'p', 'ret_neto', 'resultado']].round(4))
```
